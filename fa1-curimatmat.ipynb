```r
library(ggplot2)
library(dplyr)

cytof_data <- read.csv("/kaggle/input/datasets/zairacurimatmat/cytof-one-experiment-csv/cytof_one_experiment.csv")

base_plot <- ggplot(cytof_data, aes(x = CD16))

histogram_plot <- base_plot + 
  geom_histogram(bins = 30, fill = "dark green", color = "black") +
  xlab("CD16") +
  ylab("Count") +
  ggtitle("Histogram of CD16")

ecdf_plot <- base_plot + 
  stat_ecdf(geom = "step", color = "dark green") +
  xlab("CD16") +
  ylab("ecdf") +
  ggtitle("ECDF of CD16")

print(histogram_plot)
print(ecdf_plot)
```

```r
library(ggplot2)
library(dplyr)

cytof_data <- read.csv("/kaggle/input/datasets/zairacurimatmat/cytof-one-experiment-csv/cytof_one_experiment.csv")
qq.df = as.data.frame(qqplot(cytof_data$CD107a, cytof_data$NKG2A, plot.it = FALSE))

ggplot(qq.df, aes(x = x, y =y)) +
    geom_point(alpha = 0.3, color = "green") +
    geom_abline(slope = 1, intercept = 0, linetype = "dashed") +
    xlab("CD107a Quantiles") + ylab("NKG2A Quantiles") +
    ggtitle("Q-Q Plot: CD107a vs NKG2A")
```
