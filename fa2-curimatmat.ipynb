```r
library(tidyverse)
library(dplyr)

cytof_data <- read.csv("/kaggle/input/datasets/zairacurimatmat/cytof-one-experiment-csv/cytof_one_experiment.csv")

cytof_long <- cytof_data %>%
    pivot_longer(
        cols = everything (), 
        names_to = "Protein", 
        values_to = "Amount"
    )

summary <- cytof_long %>%
    group_by("Protein") %>%
    summarise(
        median_value = median(Amount, na.rm = TRUE),
        mad_value = mad(Amount, na.rm = TRUE)
    )

ggplot(summary, aes(x = mad_value, y = median_value)) +
    geom_point(color = "dark green", size = 4) +
    labs(title = "Protein Level Maker Spreadlocation (S-L) Plot",
         x = "MAD",
         y = "Median") +
theme_minimal()

dim(cytof_long)
head(cytof_long)
head(summary)
```
